```python
import numpy as np

f = lambda x: -x**2 + 5*x + 20          # maximize
x = np.array([-9.6, -6, -2.6, -1.1, 0.6, 2.3, 2.8, 8.3, 10])
v = np.zeros(9)
c1 = c2 = 1
r1, r2 = 0.213, 0.876                    # same values as class

pbest = x.copy()
gbest = x[np.argmax(f(x))]
stall = 0

for t in range(1, 101):
    v = v + c1*r1*(pbest - x) + c2*r2*(gbest - x)
    x = x + v
    better = f(x) > f(pbest)
    pbest[better] = x[better]
    old = gbest
    gbest = pbest[np.argmax(f(pbest))]

    if t <= 3:
        print(f"Iter {t}: x = {np.round(x, 4)}")
        print(f"   gbest = {gbest:.4f}, f = {f(gbest):.4f}\n")

    # convergence: gbest stops improving for 10 iterations
    stall = stall + 1 if abs(f(gbest) - f(old)) < 1e-6 else 0
    if stall == 10:
        print(f"Converged at iter {t}: x = {gbest:.4f}, f = {f(gbest):.4f}")
        break
```
